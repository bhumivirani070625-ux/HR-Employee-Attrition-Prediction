```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("winequality.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull()
```

```python
df.isnull().sum()
```

```python
plt.figure(figsize=(25,25))
sns.heatmap(df.isnull())
```

```python
null_var=df.isnull().sum()/df.shape[0]*100
```

```python
null_var
```

```python
drop_col=null_var[null_var>20].keys()
```

```python
drop_col
```

```python
df2_drop=df.drop(columns=drop_col)
df2_drop
```

```python
df3=df2_drop.select_dtypes(include=['int64','float64'])
```

```python
df3
```

```python
sns.heatmap(df3.isnull())
```

```python
df4=df3.fillna(df3.mean())
df4.isnull().sum().sum()
```

```python
sns.heatmap(df4.isnull())
```

```python
df4.duplicated().sum()
```

```python
df4 = df4.drop_duplicates()
```

```python
df4.duplicated().sum()
```

```python
df4.info()
```

```python
sns.boxplot(data=df4)
```

```python
Q1 = df4.quantile(0.25)
Q3 = df4.quantile(0.75)
IQR = Q3 - Q1

Q1,Q3,IQR
```

```python
upper_limit=Q3+(1.5*IQR)
lower_limit=Q1-(1.5*IQR)

upper_limit,lower_limit
```
